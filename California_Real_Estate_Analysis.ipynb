```python
import pandas as pd
import numpy as np
url="https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv"
cloud_df=pd.read_csv(url)
print(cloud_df.head())
```

```python

import matplotlib.pyplot as plt

print("BOOTING VISUAL ENGINE...")

# Command Pandas to draw a scatter plot using the GPS coordinates
cloud_df.plot(kind="scatter", x="longitude", y="latitude", alpha=0.4,
              s=cloud_df["population"]/100, label="Population", figsize=(10,7),
              c="median_house_value", cmap=plt.get_cmap("jet"), colorbar=True)

plt.title("California Real Estate Heatmap")
plt.legend()
plt.show()
```
